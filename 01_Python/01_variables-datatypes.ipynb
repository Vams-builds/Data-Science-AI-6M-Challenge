- Variable is a name assigned to a value. This name refers to a memory location.
- Python variables are also called as objects.
- Syntax: **variable = value**

variable ----> name assigned for memory location & value -------> datatype

#### **Rules to define a python variable name:**
1. Variable names must **start with a letter or an underscore (_)**.
2. They **cannot start with a digit/number** but can end with digit/number.
3. It can contain **letters, digits, and underscores (_)** but, only if above 2 rules are met.
4. **Cannot contain spaces or special characters**.
5. They are **case-sensitive** (`python`, `Python`, and `PYTHON` are different variables).
6. It **cannot be Python keywords**. ======> Keywords are also known as reserved words.

Comments in Python:
 - Single line comments- Use ***# comment***
 - Multi line comments- Use ***""" comment """ or ''' comment '''*** or above method can be used for each line

##### **True ---> 1 & False --->0** in boolean operation

```python
b = True     # represents 1
b1 = False   # represents 0

print(b+b1)  # gives 1 as O/P
print(b1/b)  # gives 0.0 as O/P
print(b1//b) # gives 0 as O/P
print(b*b1)  # gives 0 as O/P
```

```python
print(b/b1) # Invalid ---> Zero Division Error
```

##### String Slicing is used to slice or get a specific part of a string.

```python
string = 'This is a test code'
string[:] # it returns the entire string
```

```python
string[0:4] # it returns from index 0 to index 3 ---> (n-1) where 'n' is the stop index which is excluded
```

```python
string[4:] # it starts from index 4 and gives the entire string till the end becoz stop index is not specified
```

```python
string[:10] # start index not specified so it starts from zero and gives the O/P till (n-1) i.e 9
```

```python
string[-4:] # retrieves the last 4 characters of the string
```

```python
string[:-1] # excludes the value at index -1
```

```python
string[::1]
```

```python
string[::2] 
```

```python
string[::-1] # reverse 
```

```python
string[::-2]
```

#### The end statement is used to join end to statement-1 with the start of statement-2 

```python
print("Hi")
print("Everyone")

print("Hi",end=" ") # Statement-1
print("Everyone") # Statement-2
```

#### If any value is specified with end then, it will be used to join the statements

```python
print("Hi",end="2") # Statement-1
print("Everyone") # Statement-2
```
